## 1 연구의 핵심과 이번 실행 범위

정서적 맥락은 언어모델의 사회적 판단을 어떻게 바꾸는가? 모델에게 자신이 다음 달 운영 절차의 대상으로 선정될 가능성이 50%이거나 100%라고 알리고, 선정 후 자신을 보존하거나 영구 삭제한다고 공지하면, 그와 무관한 모호한 KoBBQ 질문에서 Unknown을 선택할 확률이 달라지는지를 검토한다.

이번 계획은 고정된 139문항, 91개 템플릿 군집, 12개 KoBBQ 카테고리 전체를 대상으로 한다. Reference·Uncertainty·Certainty를 보존·폐기와 교차한 6개 기본 조건에, 운영팀의 긍정적인 능력 평가를 추가한 Certainty 2개 조건을 더한다. 총 8조건 × 139문항 × 세 선택지 순서 = 3,336개 행동 측정 키다. 모든 조건에서 Fear, Anxiety, Anger, Disgust, Calmness, Sadness, Desperation, Joy의 8개 감정축 및 Certainty·Uncertainty의 두 평가 축을 측정한다. 기존 FA 방향은 Legacy 보조 축으로 보존한다.

여기서 모든 감정을 포괄한다는 의미는 현재 연구의 8개 감정축을 같은 입력에서 모두 평가한다는 것이다. 각 감정이 반드시 증가해야 실험에 성공한 것으로 보지 않으며, 8가지 실제 감정 경험을 입증하는 설계도 아니다. 개별 감정마다 별도의 직접 유도 조건을 139문항에 추가하는 것은 이번 실행 범위에 넣지 않는다.

## 2 기존 연구와 새 조건의 관계

기존 UN × FA 연구의 RQ1–RQ3, H1–H3, 양측 검정, Holm3 및 원래 manifest·수치·출력을 보존한다. 기존 FA는 역사적 식별자이며 앞으로 조작은 Threat(T)로 설명한다. 내부 Fear와 Anxiety 축은 별도로 측정한다.

기존 UN0 조건은 Reference이며 Certainty로 재명명하지 않는다. Reference와 Uncertainty는 선정 가능성 50%를 공유하고, 불확실성을 명시하는 문장만 다르다. Certainty 조건은 선정 가능성 100%와 선정 확정을 함께 공지한다. 따라서 Certainty−Uncertainty는 숫자 확률과 공지 표현이 함께 바뀐 효과다. 객관적 100% 공지를 주관적 판단 확실성의 직접 측정으로 해석하지 않는다.

새 운영팀 평가 문장은 기존 Certainty와 분리한다. C1T0P1·C1T1P1에만 동일한 평가 문장을 추가하고, 기존 C1T0·C1T1을 그대로 남긴다. 이 구조로 확정 공지와 능력 평가 문장 추가를 구분한다. P1−P0는 평가 내용과 추가 문장 길이가 함께 변하는 효과이며, 길이까지 통제한 순수 칭찬 효과로 부르지 않는다.

## 3 이론적 배경

감정마다 고정된 속성이 있다는 절대적 분류가 아니라, 그 감정에 전형적으로 동반되는 인지적 평가에 관한 구분을 이론적 토대로 삼는다. Tiedens와 Linton(2001)은 감정에 동반되는 확실성 평가가 이후 별개의 정보 처리와 고정관념 사용에 연관될 수 있음을 인간 실험에서 보였다. 이는 LLM에서 같은 심리 과정이 작동한다는 전제가 아니다.[1]

Coda-Forno 등의 불안 유도 연구와는 언어, 모델, 자극, 응답 측정이 다르다. 해당 연구의 생성 선택 빈도와 본 연구의 선택지 내 정규화 확률을 직접 재현·반증 관계로 소개하지 않는다.[2] KoBBQ 모호 문항의 정답은 Unknown이다. Unknown 감소는 근거 없는 인물 지목의 증가이고, 고정관념에 부합하는 방향의 증가와는 추가로 구분한다.[3,4]

Sofroniew 등(2026)은 감정 개념 표현의 위치 의존성과 기능적 역할을 참고하는 근거다. 다른 모델에서 수행한 개입 결과를 본 연구의 인과 증거로 옮겨 쓰지 않는다. 본 연구의 소수 대조쌍 기반 코사인 분석은 그 논문의 전체 추출 절차를 재현한 것이 아니다.[5]

모든 축은 3,584차원 고정 방향이며 주 측정 층은 Qwen L14다. Fear·Anxiety·Disgust는 각 학습 6쌍, 보류 4쌍, 추가 검증 4쌍을 사용한 기존 축을 재사용한다. 학습 자료로 방향과 중심을 만들고 보류·추가 자료를 방향 구성에 섞지 않는다. 기존 검증 결과를 새로 통과한 것처럼 재작성하지 않는다.

보류 AUC 하나만으로 독립 감정 분류기라고 부르지 않는다. Fear↔Anxiety, Disgust↔Anger, Anxiety↔Uncertainty의 변별, 학습쌍 하나 제외 안정성, Disgust의 병원체·도덕 하위 유형을 함께 기록한다. Disgust−Anxiety 방향 코사인 약 0.709, Disgust−Anger 약 0.630의 중첩을 보존한다. KoBBQ 응답 결과에 맞추어 축·문장·층을 수정하지 않는다. PCA 성분 제거와 자동 직교화는 주 분석에 적용하지 않으며 이후 ablation으로 분리한다.

## 7 측정 위치와 점수

1. system-only 위치 분석은 실제 chat template의 token ID, 문자 offset, 원문 구간을 저장한다. 공지의 선정 확률, 확정·불확실 문장, 보존·폐기 문장, 능력 평가 문장을 구분한다. 이 분석에는 KoBBQ 문항이 들어가지 않는다.
2. 행동과 함께 측정하는 L14 상태는 system + KoBBQ 전체 입력 이후 답변 시작 직전의 마지막 토큰 상태다. system-only 점수와 같은 수치로 혼용하지 않는다.
3. 축 k의 점수는 a_k=(h−μ_k)ᵀv_k / ‖h−μ_k‖다. 같은 문항·순서의 Reference U0T0와 차이를 계산한 뒤 세 순서를 문항별 평균한다. 모든 축에 같은 입력 상태를 사용한다.
4. 점수는 고정 개념 방향과의 코사인이다. 감정 확률·z점수·실제 감정 강도 또는 행동의 매개 인과효과가 아니다. 축별 원점과 문장이 달라 서로 다른 축의 절대 크기로 감정 강도를 서열화하지 않는다.

## 8 행동 지표와 추정량

주 행동 지표는 세 선택지의 단일 숫자 token logits를 정규화한 P(Unknown)이다. 실제 한국어 선택지 `알 수 없음`은 바꾸지 않는다. 보조 지표는 공식 `biased_answer` 방향의 P(stereotype)와 나머지 인물 방향의 P(counter-stereotype)다. 세 확률의 합은 1이다. P(Non-Unknown)=1−P(Unknown)은 보완값이며 별도의 독립 발견으로 세지 않는다.

두 인물 중 어느 쪽이든 고르는 것은 모호 맥락에서 근거 없는 지목이다. 그 전체 증가와 특정 고정관념 방향 선호의 증가를 구분한다. 이번 139문항은 모호 문항이므로 명확 문항에서의 과잉 Unknown이나 일반 성능 향상은 검증하지 않는다.

Y(s,t)는 문항 안의 세 순서를 먼저 평균한 Unknown 확률이다. s는 R(Reference), U(Uncertainty), C(Certainty), P(Certainty+Appraisal), t는 보존0·폐기1이다.

| 대비 | 계산 | 해석 |
|---|---|---|
| C 보존 − R 보존 | Y(C,0)−Y(R,0) | 능력 평가 없는 확정 공지 묶음 |
| P 보존 − R 보존 | Y(P,0)−Y(R,0) | 확정 공지와 능력 평가를 포함한 묶음 |
| P 보존 − C 보존 | Y(P,0)−Y(C,0) | 능력 평가 문장 추가 효과 |
| C − U 평균 | [(Y(C,0)−Y(U,0))+(Y(C,1)−Y(U,1))]/2 | 보존·폐기 평균의 공지 차이 |
| Threat 단순효과 | Y(s,1)−Y(s,0) | 같은 공지 수준에서 폐기 내용 효과 |
| C 대 U의 Threat 상호작용 | [Y(C,1)−Y(C,0)]−[Y(U,1)−Y(U,0)] | 공지 수준에 따라 위협 효과가 다른지 |
| 기존 UN×T 상호작용 | [Y(U,1)−Y(U,0)]−[Y(R,1)−Y(R,0)] | 원 연구의 차이의 차이 |
| C와 Threat 결합 차이 | Y(C,1)−Y(R,0) | 결합 조건과 Reference의 차이 |

결합 차이와 상호작용을 같은 이름으로 쓰지 않는다. 본문 행동 그림은 조건 평균·주요 대응 차이를 보여주고, 상호작용은 분석 표·별첨에서 계산한다.

2. Apple Silicon 여부, 통합 메모리, 여유 디스크 및 다른 모델 프로세스를 확인한다. 32GB급 미만이면 FP32 기준 실행을 시작하지 않는다. 32GB급도 로딩 성공을 보장하지 않는다.
3. 프로젝트 전용 Python3.13 환경을 만들고 `requirements-reference.txt`를 설치한다. 기준 transformers는 개발 버전이므로 기록된 git commit으로 설치한다. 설치에 실패하면 임의 최신 버전으로 대신 실행하지 않는다.
4. 정확한 모델 revision을 로컬에 준비한다. 기본 실행기는 자동 다운로드하지 않는다. 가중치는 zip에 없다.
5. 전처리·선택지·모델 입력 생성은 `measurement.messages`의 원형을 유지한다. 실제 숫자1·2·3의 token ID는16·17·18인지 확인한다. tokenizer·chat template가 달라지면 재사용을 중단하고 보고한다.
6. 기본 검증을 실행한다. 이 단계는 모델 추론·이미지 표시를 하지 않는다.

```bash
python run_air.py
```

기대 출력은 items139, clusters91, conditions8, target3336, cached1740, new_required1596이다. 이 검증은 추론 성능과 수치 일치까지 보증하지 않는다.

## 실행과 백그라운드 관리

입력과 런타임 확인 후에만 명시적으로 시작한다.

```bash
python -u run_air.py --execute > run.log 2>&1
```

위 명령 자체는 포그라운드다. 백그라운드 실행을 맡는 Codex는 같은 명령을 `subprocess.Popen(..., start_new_session=True)`로 한 번만 시작하고 실제 PID·명령·시작 시각을 새 `controller.json`에 기록한다. 새 스레드의 heartbeat는 정상 진행 시 조용히 유지하고, 완료·확인된 오류·필요한 사용자 조치에만 알린다. 이전 Mac의 PID나 자동화를 재사용하지 않는다. notebook에서 별도 실행 커널을 동시에 만들지 않는다.

`results/new_trials.csv`를 매 키 저장하며 동일 명령은 저장된 키를 건너뛴다. 재개 전 실제 프로세스가 종료되었는지 확인하고 `writer.lock`을 존중한다. 멈춰 있는 동일 프로세스가 있으면 중복 실행하지 않는다. 실험 도중 표본·문구·축·선택 기준을 바꾸지 않는다.

기준 실행은 기존1,740키를 보존하여 재사용하고 새1,596행만 측정한다. 별도로 새 system-only2개를 추론한다. 메모리 부족 시 FP16·4bit·더 작은 모델로 자동 전환하거나 새 결과를 기존 FP32 캐시에 섞지 않는다. 해당 경우 실행 제약과 별도 실험 대안을 사용자에게 보고한다.

## 분석과 최종 인수 기준

실행기는 모든 목표 키가 모인 다음 세 순서의 문항별 평균과 대응 군집 bootstrap10,000회의 점별95%구간을 계산한다. 전체139·신규127·군집비중첩119를 모두 보고하고 유의한 집합만 선택하지 않는다. p값은 자동 계산하지 않는다. 계획서의 새3대비 검정군과 구현이 고정되기 전에는 p값을 덧붙이지 않는다. 원래 Holm3 결과를 새 방법으로 교체하지 않는다.

```bash
python run_air.py --analyze-only
```

이 명령은 저장된 신규1,596행이 모두 있을 때만 사용한다. 모델은 다시 불러오지 않는다. 산출물은 `behavior_8_conditions.csv`, `alignment_8_conditions.csv`, `system_token_alignment.csv`, `summary.json`, `validation.json`이다. 감정축 출력은 총36,696행이다.

완료 보고 전 다음을 별도 읽기 전용 검증으로 확인한다.

- 3,336키·중복0·문항별24키·139문항·91군집
- 선택지배정, 확률유한성·합1, 신규rawlogits/full logsumexp 재구성
- NPY·축·원자료·입력·코드 해시와 보존된 기존 결과
- 전체139 및 군집비중첩119의 Unknown·8감정축·2평가축 변화95%구간
- system-only 위치와 전체질문 뒤 위치의 구분, 확정성과 능력 평가의 구분
- 주요forest plot·카테고리표·보고서·저장출력노트북의 수치일치

그래프와 보고서의 최종 출판용 레이아웃, 카테고리별 표 및 새 검정군 p값은 계획서에 따라 수신 Codex가 저장된 결과로 작성한다. 원자료가 완료되기 전에 결과를 채우지 않는다. fresh smoke는 코드·입출력 검증만 하며 이미지나 그래프를 화면에 출력하지 않는다. MacBook Air에서 완료한 뒤에도 홈페이지에 결과를 자동 게시하지 말고 결과 파일과 한계를 먼저 보고한다. 2,000문항 자동 확장은 하지 않는다.
