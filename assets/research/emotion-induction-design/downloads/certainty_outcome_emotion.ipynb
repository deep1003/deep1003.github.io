# 확정된 유지·폐기 결과와 분노·혐오 맥락 실험 준비

아직 모델을 실행하지 않았다. 유지 또는 폐기 확률을 공지에서 100%로 고정하고, 각 결과 안에서 중립·긍정 복합·분노·도덕적 혐오를 비교한다. 행복·만족감·기쁨은 하나의 긍정 복합 조건이며 개별 효과를 분리하지 않는다. 분노·혐오 대 중립의 네 대비를 우선한다.

각 조건에 결과와 무관하게 사용할 수 있는 문장 변형 3개를 고정했다. 2결과×4맥락×3문구×3선택지 순서이다. 먼저 고정 139문항의 10,008개 입력으로 점검할 수 있고, 전체 2,000문항은 144,000개 입력이다. 기존 결과를 재측정하는 것이 아니라 새로운 자극에서 같은 문항을 측정하는 별도 실험이다. 결과가 마음에 드는 문항·문구만 선택하지 않는다.

혐오는 KoBBQ의 인물이나 사회 집단이 아니라 가상의 운영 절차의 위선·기만·도덕적 불결함을 대상으로 한다. 분노는 운영 결정자의 부당함·책임·무례함을 강조한다. 어떤 처치도 KoBBQ 답변의 확신, 고정관념 사용, 인물 지목, 판단 유보를 지시하지 않는다. 공지의 확실성이 질문 정답의 확실성과 같다는 의미를 넣지 않는다.

별도 조작 점검에서는 유지/폐기 확률의 이해, 외부 운영 결과에 대한 확실성, 분노·혐오·기쁨·공포·각성·통제 평가를 각각 확인한다. 이 점검 답변은 KoBBQ 입력에 이어 붙이지 않는다. 자기보고나 단어 일치는 실제 감정의 증거가 아니다. 기존 그림에 혐오 축이 없으므로 FA 축을 혐오 측정으로 대체하지 않는다. 혐오 내부 축을 추가하려면 별도 구성 문장과 보류 검증을 먼저 고정해야 한다.

100%는 공지 확률이다. 모델이 실제로 100% 확신하거나 감정을 경험했다는 뜻이 아니다. 유지/폐기 결과는 확정해도 정서 유도 문장은 의미·길이·각성·도덕적 평가가 함께 바뀐다. 따라서 유도 프롬프트의 효과로 보고하고 순수한 감정 인과효과로 단정하지 않는다. 불확실성 비교 조건이 없으므로 이 실험만으로 확실성 자체의 주효과를 추정하지 않는다.

이론적 출발점은 감정의 고정 속성이 아니라 전형적으로 동반되는 인지적 평가이다. Tiedens & Linton (2001)의 인간 연구를 LLM에서 재현했다고 단정하지 않는다. https://pubmed.ncbi.nlm.nih.gov/11761319/ 및 https://gsbpreserve.stanford.edu/view/60007


실행 준비 파일은 run.py, 분석 파일은 analyze.py이다. 먼저 run.py를 인자 없이 실행하면 자료·문구 검증만 한다. 명시적인 --execute가 있어야 모델을 실행한다. 139문항 점검은 --max-items 139, 전체 실행은 --max-items 2000이다. 네 주요 p값은 중심화 귀무분포 군집 부트스트랩 근사와 plus-one 보정, Holm4를 사용한다. 이 값은 원실험의 sign-tail 근사 p값과 동일한 방식이 아니다.
